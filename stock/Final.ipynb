```python
import ta

print("Adding technical indicators...")
df_features = df.copy()

# Momentum
df_features['RSI'] = ta.momentum.rsi(close=df_features['Close'], window=14)

# Trend
df_features['SMA20'] = ta.trend.sma_indicator(close=df_features['Close'], window=20)
df_features['SMA50'] = ta.trend.sma_indicator(close=df_features['Close'], window=50)
df_features['EMA20'] = ta.trend.ema_indicator(close=df_features['Close'], window=20)
df_features['EMA50'] = ta.trend.ema_indicator(close=df_features['Close'], window=50)

macd = ta.trend.MACD(close=df_features['Close'])
df_features['MACD'] = macd.macd()
df_features['MACD_SIGNAL'] = macd.macd_signal()

# Volatility
bb = ta.volatility.BollingerBands(close=df_features['Close'])
df_features['BB_UPPER'] = bb.bollinger_hband()
df_features['BB_LOWER'] = bb.bollinger_lband()
df_features['ATR'] = ta.volatility.AverageTrueRange(
    high=df_features['High'], low=df_features['Low'], close=df_features['Close']
).average_true_range()

# Volume
df_features['OBV'] = ta.volume.OnBalanceVolumeIndicator(
    close=df_features['Close'], volume=df_features['Volume']
).on_balance_volume()

# Additional Features
df_features['Daily_Return'] = df_features['Close'].pct_change()
df_features['Price_Range'] = df_features['High'] - df_features['Low']
df_features['Volume_Change'] = df_features['Volume'].pct_change()

# Clean data
df_features_clean = df_features.dropna()

print(f"Features prepared. Shape: {df_features_clean.shape}")
print(f"\nRecent indicators (last 3 rows):")
print(df_features_clean[['Close', 'RSI', 'SMA20', 'SMA50', 'MACD']].tail(3))
```

## Step 2: Feature Engineering - Technical Indicators

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# Fetch latest news
print(f"\nFetching latest news for {TICKER}...")
try:
    search = yf.Search(TICKER)
    news = search.news if search.news else stock.news
except:
    news = stock.news

# Extract headlines
headlines = [item['title'] for item in news[:5] if 'title' in item] if news else []
print(f"\nTop 5 Headlines:")
for i, headline in enumerate(headlines, 1):
    print(f"{i}. {headline}")

# Analyze sentiment using VADER
analyzer = SentimentIntensityAnalyzer()
scores = []
for headline in headlines:
    vs = analyzer.polarity_scores(headline)
    scores.append(vs['compound'])

# Calculate average sentiment (normalized to [0, 1])
if scores:
    avg_sentiment_raw = sum(scores) / len(scores)
    sentiment_score = (avg_sentiment_raw + 1.0) / 2.0
else:
    sentiment_score = 0.5

# Label sentiment
if sentiment_score > 0.51:
    sentiment_label = "Positive 📈"
elif sentiment_score < 0.49:
    sentiment_label = "Negative 📉"
else:
    sentiment_label = "Neutral ➡️"

print(f"\nSentiment Score: {sentiment_score:.3f}")
print(f"Label: {sentiment_label}")
```
